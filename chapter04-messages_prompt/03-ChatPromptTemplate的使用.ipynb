# ChatPromptTemplate的使用

## 1.ChatPromptTemplate的实例化

### 1.1 from_messages() -- 推荐使用

举例：

```python
from langchain_core.prompts import ChatPromptTemplate
from rich import print as rprint

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "你是一个友好的AI助手，你的名字叫做{name}"),
    ("user", "你好，我喜欢你"),
    ("ai", "我也喜欢你，谢谢"),
    ("user", "{user_input}")
])

# 调用
res = chat_prompt_template.invoke({"name":"小园", "user_input": "2 + 2 = ?"})
rprint(res)
```

## 2.ChatPromptTemplate的调用

### 2.1 invoke()

传入参数：字典列表；返回值类型：ChatPromptValue

举例：

```python
from langchain_core.prompts import ChatPromptTemplate
from rich import print as rprint

chat_prompt_template = ChatPromptTemplate([
    ("system", "你是一个友好的AI助手，你的名字叫做{name}"),
    ("user", "你好，我喜欢你"),
    ("ai", "我也喜欢你，谢谢"),
    ("user", "{user_input}")
])

# 调用
res = chat_prompt_template.invoke({"name":"小园", "user_input": "2 + 2 = ?"})
rprint(res)
rprint(type(res))
```

### 2.2 format()

参数类型：变量值；返回值：字符串

举例：

```python
from langchain_core.prompts import ChatPromptTemplate
from rich import print as rprint

chat_prompt_template = ChatPromptTemplate([
    ("system", "你是一个友好的AI助手，你的名字叫做{name}"),
    ("user", "你好，我喜欢你"),
    ("ai", "我也喜欢你，谢谢"),
    ("user", "{user_input}")
])

# 调用
res = chat_prompt_template.format(name = "小园", user_input = "2 + 2 = ?")
rprint(res)
rprint(type(res))
```

### 2.3 format_messages()

传入参数：变量的值 返回值：list

举例：

```python
from langchain_core.prompts import ChatPromptTemplate
from rich import print as rprint

chat_prompt_template = ChatPromptTemplate([
    ("system", "你是一个友好的AI助手，你的名字叫做{name}"),
    ("user", "你好，我喜欢你"),
    ("ai", "我也喜欢你，谢谢"),
    ("user", "{user_input}")
])

# 调用
res = chat_prompt_template.format_messages(name = "小园", user_input = "2 + 2 = ?")
rprint(res)
rprint(type(res))
```

## 3.结合大模型的调用

举例：

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
from langchain_core.prompts import ChatPromptTemplate
from rich import print as rprint


# 提供大模型
load_dotenv(override=True)

MODEL_NAME = "deepseek-flash"

model = init_chat_model(
    model=MODEL_NAME,
    model_provider="deepseek",
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
)

# 提供提示词模板
chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "你是一个友好的AI助手，你的名字叫做{name}"),
    ("user", "你好，我喜欢你"),
    ("ai", "我也喜欢你，谢谢"),
    ("user", "{user_input}")
])

prompt_value = chat_prompt_template.invoke({"name":"小园", "user_input": "2 + 2 = ?"})


# 模型调用
res = model.invoke(prompt_value)
rprint(res)
rprint(type(res))
```
