# TypeDict格式的使用

## 1.理解

```python
from typing_extensions import TypedDict

class MovieDict(TypedDict):
    title: str
    year: int
    director: str
    rating: float

movie: MovieDict = {
    "title": "盗梦空间",
    "year": 2010,
    "director": "克里斯托弗.诺兰",
    "rating": 8.8
}

print(movie)
```

## 2.基本使用

举例1:

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

load_dotenv(override = True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    model = "deepseek-flash",
    model_provider = "deepseek",
    api_key = DEEPSEEK_API_KEY,
    base_url = DEEPSEEK_BASE_URL,
        extra_body = {
        "thinking":{"type":"disabled"}
    }
)
```

```python
from typing_extensions import Annotated, TypedDict

class MovieTypeDict(TypedDict):
    """电影的信息"""
    title: Annotated[str, "电影的名称"]
    year: Annotated[int, "电影的上映时间,四位数"]
    director: Annotated[str, "电影的导演"]
    rating: Annotated[float, "电影的评分，满分10分，可以包含一位小数"]


structured_model = model.with_structured_output(MovieTypeDict)
res = structured_model.invoke("给我介绍一下电影《红高粱》")
print(res)
print(type(res))
```

举例2：嵌套结构的使用

```python
from typing_extensions import Annotated, TypedDict
from typing import List

class Actor(TypedDict):
    """演员的信息"""
    name: Annotated[str, "演员的名字"]
    role: Annotated[str, "扮演的角色"]

class MovieTypeDict(TypedDict):
    """电影的信息"""
    title: Annotated[str, "电影的名称"]
    year: Annotated[int, "电影的上映时间,四位数"]
    director: Annotated[str, "电影的导演"]
    rating: Annotated[float, "电影的评分，满分10分，可以包含一位小数"]
    cast: Annotated[List[Actor], "演员的列表"]   # 嵌套列表的结构

structured_model = model.with_structured_output(MovieTypeDict)
res = structured_model.invoke("给我介绍一下电影《红高粱》")
print(res)
print(type(res))
```
