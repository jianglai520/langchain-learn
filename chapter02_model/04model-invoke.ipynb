```python
from langchain.chat_models import init_chat_model
import os
from dotenv import load_dotenv
from langchain_core.messages import SystemMessage, HumanMessage,AIMessage

# 加载配置文件
load_dotenv(override = True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 获取大模型
model = init_chat_model(
    model = "deepseek:deepseek-flash",
    api_key = DEEPSEEK_API_KEY,
    base_url = DEEPSEEK_BASE_URL,
)

message = [
    SystemMessage(content = "你是一名专业的数学老师"),
    HumanMessage(content = "给我讲讲导数是啥？")
]

res1 = model.invoke(message)
print(f"第一次对话：{res1.content}")

message.append(AIMessage(content = res1.content))   # content一定要加.content,不然报错
message.append(HumanMessage(content = "我上一个问你的是啥问题？"))

res2 = model.invoke(message)
print(f"第二次对话:{res2.content}")
```

## 2.invoke的返回值

举例1：

```python
from langchain.chat_models import init_chat_model
import os
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage

# 加载配置文件
load_dotenv(override = True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 获取大模型
model = init_chat_model(
    model = "deepseek:deepseek-flash",
    api_key = DEEPSEEK_API_KEY,
    base_url = DEEPSEEK_BASE_URL,
)

res = model.invoke([HumanMessage(content = "2 + 3 * 3 = ?")])

print(type(res))   # <class 'langchain_core.messages.ai.AIMessage'>

print(res)
```

```python
res = model.invoke("用一句话解释什么是 AI")

# 1.获取回复内容
print("AI 回复:", res.content)

# 2.获取响应元数据
metadata = res.response_metadata
print(f"使用的模型: {metadata['model_name']}")
print(f"结束原因: {metadata['finish_reason']}")
print(f"模型提供商: {metadata['model_provider']}")

# 3.获取 Token 使用情况
usage = metadata.get('token_usage', {})
print(f"输入 Tokens: {usage['prompt_tokens']}")
print(f"输出 Tokens: {usage['completion_tokens']}")
print(f"总计 Tokens: {usage['total_tokens']}")

# 4.获取消息 ID
print(f"消息 ID：{res.id}")
```
